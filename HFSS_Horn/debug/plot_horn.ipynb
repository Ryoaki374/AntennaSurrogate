# 円柱形状の作成と表示
`n_vertices=None` では真円、3以上の整数ではその円に内接する正多角形を断面に使います。長さの単位は mm です。

```python
from pathlib import Path

from lib_RFdesign_debug import Cylinder
```

## 真円の円柱

```python
cylinder = Cylinder(Path("cylinder.step"))
circle_points = cylinder.genCylinder(diameter=10.0, height=20.0)
cylinder.plotCylinder3D(circle_points);
```

## 内接する正多角形の柱（例: 正六角形）

```python
polygon = Cylinder(Path("hexagonal_cylinder.step"))
polygon_points = polygon.genCylinder(diameter=10.0, height=20.0, n_vertices=6)
polygon.plotCylinder3D(polygon_points);
```
